## 문제 요약

- `(의상 이름, 의상 종류)` 형태의 배열 `clothes`가 주어진다
- 하루에 각 종류별로 **최대 1개만 착용 가능**
- **최소 1개 이상**은 반드시 착용해야 한다

서로 다른 의상 조합의 개수를 반환한다.

## 함정 / 실수 포인트

### ❗ 경우의 수를 직접 나열하려 하면 복잡해짐

- 조합을 직접 만들면 경우의 수가 급격히 증가
- 핵심은 **수학적 조합 공식**

### ❗ 안 입는 경우를 고려해야 함

- 각 종류별 선택지는 `(입는다 + 안 입는다)`
- 단, **모두 안 입는 경우는 제외**

## 활용한 개념

- 딕셔너리를 이용한 종류별 분류
- 경우의 수 곱셈 원리
- `len(v) + 1` → 해당 종류를 안 입는 경우 포함

```python
def solution(clothes):
    item_dic = {}
    for item in clothes:
        if item[1] not in item_dic:
            item_dic[item[1]] = 0
        item_dic[item[1]] += 1

    answer = 1
    for v in item_dic.values():
        answer *= (v + 1)

    return answer - 1
```

## 한 줄 정리

> 이 문제의 핵심은 **종류별 (개수 + 안 입는 경우)를 곱한 뒤, 모두 안 입는 경우 1을 빼는 것**이다.
