# 📒 폰켓몬 – 오답 노트

프로그래머스 문제: **폰켓몬**

## 문제 요약

- 폰켓몬 번호가 담긴 배열 `nums`가 주어진다
- 전체 개수의 **절반만 선택 가능**
- **서로 다른 종류를 최대한 많이 선택**해야 한다

선택할 수 있는 최대 폰켓몬 종류 수를 반환한다.



## 함정 / 실수 포인트

### ❗ 나눗셈 결과 타입

- `len(nums) / 2` → **float 반환**
- 문제는 **정수 반환**을 요구함

➡️ 반드시 `//` 또는 `int()` 처리 필요

### ❗ 조건 분기 자체는 필요 없음

- 핵심은 `min(종류 수, 선택 가능 수)`


## 활용한 개념

- `set()`을 이용한 중복 제거
- `len()`으로 개수 계산
- `min()`을 이용한 조건 단순화


```python
def solution(nums):
    return min(len(set(nums)), len(nums) // 2)

def solution(nums):
    if len(nums)/2 > len(set(nums)):
        return len(set(nums))
    else:
        return len(nums)/2
```

## 한 줄 정리

> 이 문제의 핵심은 **최대 선택 수와 종류 수 중 더 작은 값**이며, `/`와 `//`의 차이를 놓치면 오답이 된다.
