## 문제 요약

- 참가자 배열 `participant`
- 완주자 배열 `completion`
- 단 **한 명만 완주하지 못함**

완주하지 못한 선수의 이름을 반환한다.


## 함정 / 실수 포인트

### ❗ 중복 이름 처리

- 선수 이름은 **중복될 수 있음**
- 단순 `set` 차집합은 사용 불가

### ❗ 마지막 한 명 예외

- 앞에서 모두 같으면
- **participant의 마지막 값이 정답**

## 활용한 개념

- 리스트 정렬 (`sort()`)
- `zip()`을 이용한 병렬 비교
- 사전순 정렬 후 불일치 지점 탐색


```python
def solution(participant, completion):
    participant.sort()
    completion.sort()

    for p, c in zip(participant, completion):
        if p != c:
            return p

    return participant[-1]
```

## 한 줄 정리

> 이 문제의 핵심은 **중복 이름을 고려한 정렬 후 비교**였다.
