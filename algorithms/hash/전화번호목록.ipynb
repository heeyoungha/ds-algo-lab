# 📒 전화번호 목록 – 오답 노트

프로그래머스 문제: **전화번호 목록**

## 문제 요약

- 전화번호 문자열 배열 `phone_book`이 주어진다
- 어떤 번호가 다른 번호의 **접두어**이면 `False`
- 그렇지 않으면 `True`를 반환한다

## 함정 / 실수 포인트

### ❗ 모든 쌍 비교는 불필요

- 이중 반복문으로 전부 비교하면 **시간 초과 위험**

### ❗ 정렬 후 인접 비교가 핵심

- 문자열 정렬 시 접두어 관계인 번호는 **서로 인접**하게 배치됨
- 따라서 **이웃한 두 값만 비교하면 충분**

## 활용한 개념

- 문자열 정렬 (`sort()`)
- `zip(a, a[1:])`을 이용한 인접 요소 비교
- 문자열 메서드 `startswith()`

```python
def solution(phone_book):
    phone_book.sort()

    for a, b in zip(phone_book, phone_book[1:]):
        if b.startswith(a):
            return False

    return True
```

## 한 줄 정리

> 이 문제의 핵심은 **정렬 후 인접한 번호만 접두어 검사**하는 것이다.
