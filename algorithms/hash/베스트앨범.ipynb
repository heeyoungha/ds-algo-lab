## 문제 요약

- 노래 장르 배열 `genres`
- 노래 재생 수 배열 `plays`
- 장르별로 **총 재생 수가 많은 순서**로 앨범 수록
- 각 장르에서 **재생 수가 많은 노래 최대 2곡** 선택
- 재생 수가 같으면 **고유 번호가 작은 노래 우선**

베스트앨범에 수록될 노래의 고유 번호를 순서대로 반환한다.

## 함정 / 실수 포인트

### ❗ 장르 정렬 기준과 노래 정렬 기준이 다름

- 장르: **총 재생 수 기준**
- 노래: **개별 재생 수 → 고유 번호**

➡️ 하나의 정렬 기준으로 처리하려 하면 구조가 꼬임

### ❗ 인덱스를 반드시 유지해야 함

- 결과는 노래 번호(index)를 반환
- `(재생 수, 인덱스)` 또는 `(인덱스, 재생 수)` 형태로 저장 필요

## 활용한 개념

- 딕셔너리 2개 사용 (장르 합계 / 장르별 노래 목록)
- `dict.get()`을 이용한 누적 합
- 다중 정렬 기준 (`lambda x: (-x[1], x[0])`)
- `enumerate()`로 인덱스 유지

```python
def solution(genres, plays):
    genre_dic = {}
    genre_song_dic = {}
    
    for i, genre in enumerate(genres):
        genre_dic[genre] = genre_dic.get(genre, 0) + plays[i]
        if genre not in genre_song_dic:
            genre_song_dic[genre] = []
        genre_song_dic[genre].append((i, plays[i]))
    
    sorted_genre_dic = sorted(
        genre_dic.items(), key=lambda x: x[1], reverse=True
    )
    
    answer = []
    for genre, _ in sorted_genre_dic:
        songs = sorted(
            genre_song_dic[genre], key=lambda x: (-x[1], x[0])
        )
        answer.extend([idx for idx, _ in songs[:2]])
    
    return answer
```
