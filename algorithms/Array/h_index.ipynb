## 문제 요약

연구자의 논문 인용 횟수 배열 `citations`가 주어진다.

- H-index는 `h`편 이상의 논문이 각각 `h`회 이상 인용된 최대 `h`
- 이를 계산하여 반환한다.

## 함정 / 실수 포인트

### ❗ 반복문이 끝까지 도는 경우

- `i + 1 > citations[i]` 조건이 **한 번도 만족되지 않을 수 있음**
- 이 경우 `return`이 실행되지 않아 오답 발생

### ❗ 마지막 예외 처리 누락

- 모든 논문이 조건을 만족하면
- H-index는 **전체 논문 수**가 됨

```python
return len(citations)
```

## 활용한 개념

- 내림차순 정렬
- `enumerate`를 이용한 인덱스 + 값 순회
- H-index 정의에 따른 조건 비교 (`i+1` vs `citations[i]`)
- 루프 종료 후 예외 처리

````python
def solution(citations):
    citations.sort(reverse=True)
    for i, item in enumerate(citations):
        if i + 1 > item:
            return i
    return len(citations)
````

## 한 줄 정리

> 이 문제의 핵심은 **반복문이 끝까지 도는 경우의 예외 처리**였다.
