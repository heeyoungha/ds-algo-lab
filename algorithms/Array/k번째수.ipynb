## 문제 요약

정수 배열 `array`와 `commands`가 주어진다.

각 `command = [i, j, k]`에 대해:

* `array`의 **i번째부터 j번째까지** 자른다
* 정렬한다
* **k번째 값**을 선택한다

각 결과를 배열로 반환한다.


## 실수 포인트

### ❗ 0-based vs 1-based 인덱스

* 문제 기준: **1-based index**
* Python 리스트: **0-based index**

따라서 반드시 다음 변환이 필요하다.

```
i번째 → i-1
j번째 → j (slice 특성상 포함됨)
k번째 → k-1
```

이 변환을 놓치면 오답.


## 활용한 개념

* Python 리스트 slice (`a:b`는 b 미포함)
* `sorted()` (원본 배열 변경 없음)
* 리스트 컴프리헨션
* 다중 변수 언패킹 (`for i, j, k in commands`)


## 최종 풀이


````python
def solution(array, commands):
    return [
        sorted(array[i-1:j])[k-1]
        for i, j, k in commands
    ]
````

## 한 줄 정리

> 이 문제의 핵심은 알고리즘이 아니라 **인덱스 기준 변환(1-based → 0-based)** 이었다.
