## 문제 요약

정수 배열 `numbers`가 주어진다.

- 각 숫자를 이어 붙여 만들 수 있는 **가장 큰 수**를 문자열로 반환한다.
- 숫자의 크기 비교가 아니라 **문자열 결합 결과 비교**가 핵심이다.

## 함정 / 실수 포인트

### ❗ 숫자 비교가 아니라 문자열 결합 비교

- 단순히 숫자 크기로 정렬하면 오답
- `x+y` 와 `y+x` 중 **더 큰 쪽이 앞에 오도록 정렬**해야 함

### ❗ 모든 값이 0인 경우

- 정렬 결과가 `'000...'` 이 될 수 있음
- 이 경우 정답은 `'0'`

```python
set(numbers) == {0}
```

## 활용한 개념

- `cmp_to_key` 를 이용한 커스텀 정렬
- 문자열 결합 비교 (`x + y` vs `y + x`)
- Python 정렬 규칙 (음수 / 양수 / 0 반환)
- 예외 케이스 처리 (all zero)

````python
from functools import cmp_to_key

def compare_two(x, y):
    if x + y > y + x:
        return -1
    elif x + y < y + x:
        return 1
    else:
        return 0

def solution(numbers):
    nums = list(map(str, numbers))
    nums.sort(key=cmp_to_key(compare_two))
    return '0' if set(numbers) == {0} else ''.join(nums)
````
