## 문제 요약

- 명함 크기 배열 `sizes` (각 원소는 [가로, 세로])
- 모든 명함을 수납할 수 있는 **지갑의 최소 크기** 구하기
- 명함은 **회전 가능** (가로와 세로를 바꿀 수 있음)

지갑의 가로 길이와 세로 길이를 곱한 값을 반환한다.

## 함정 / 실수 포인트

### ❗ 명함을 회전할 수 있다는 점을 놓치면 안 됨

- 각 명함의 가로/세로는 고정이 아님
- 모든 명함을 **일관된 방향**으로 정렬하면 최적해를 찾을 수 있음

### ❗ 전체 명함을 어떻게 배치할지 고민하면 복잡해짐

- 각 명함을 `[작은 값, 큰 값]` 형태로 통일
- 그 후 **작은 값들의 최댓값 × 큰 값들의 최댓값**이 답

➡️ 명함 배치 문제가 아니라 **정렬 + 최댓값 추출** 문제

## 활용한 개념

- 리스트 컴프리헨션으로 각 명함 정렬
- `sorted()` 함수로 각 쌍을 오름차순 정렬
- `max()`와 인덱싱을 활용한 차원별 최댓값 추출
- **그리디 접근**: 각 명함을 독립적으로 회전시켜 전체 최적해 도출

```python
def solution(sizes):
    sizes = [sorted(pair) for pair in sizes]
    return max([item[0] for item in sizes]) * max([item[1] for item in sizes])
```

## 한 줄 정리

> 이 문제의 핵심은 **모든 명함을 [작은값, 큰값] 형태로 통일한 뒤, 각 차원의 최댓값을 구하는 것**이다.
