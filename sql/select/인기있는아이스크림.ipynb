## 문제 요약

- 테이블: `FIRST_HALF`
- 조회 컬럼: `FLAVOR`
- 정렬 조건:
  1. `TOTAL_ORDER` 기준 내림차순
  2. 값이 같으면 `SHIPMENT_ID` 기준 오름차순

정렬 조건에 맞게 결과를 조회한다.

## 함정 / 실수 포인트

### ❗ ORDER BY 다중 조건 문법

- 정렬 기준이 2개 이상일 경우 **콤마(,)** 필수

❌ 잘못된 예
```sql
ORDER BY TOTAL_ORDER DESC SHIPMENT_ID ASC
```

✅ 올바른 예
```sql
ORDER BY TOTAL_ORDER DESC, SHIPMENT_ID ASC
```

## 사용 개념 정리

- `ORDER BY`는 **왼쪽 기준부터 우선 적용**됨
- 다중 정렬 시 각 기준은 **콤마로 구분**
- `ASC` / `DESC`는 각 컬럼마다 지정 가능

````python
SELECT FLAVOR
FROM FIRST_HALF
ORDER BY TOTAL_ORDER DESC, SHIPMENT_ID ASC;
````

## 한 줄 정리

> ORDER BY에서 **정렬 기준이 2개 이상이면 콤마로 반드시 구분해야 한다.**
