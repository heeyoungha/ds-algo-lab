## 문제 요약

- 테이블: `PATIENT`
- 조회 컬럼: `PT_NAME`, `PT_NO`, `GEND_CD`, `AGE`, `TLNO`
- 조건:
  - `AGE`가 12 이하
  - `GEND_CD`가 'W'
- TLNO 컬럼의 값이 `NULL`, `''`, `'   '`인 경우 `'NONE'`으로 출력
- 정렬 조건:
  1. `AGE` 내림차순
  2. `PT_NAME` 오름차순

## 함정 / 실수 포인트

### ❗ TLNO 처리 시 주의

- TLNO 값이 단순히 `NULL`만 있는 것이 아님
- 공백(`'   '`)이나 빈 문자열(`''`)도 고려해야 함

❌ 잘못된 예
```sql
SELECT PT_NAME, PT_NO, GEND_CD, AGE, COALESCE(TLNO, 'NONE') AS TLNO
FROM PATIENT
WHERE AGE <= 12 AND GEND_CD = 'W';
```
- 이 경우 공백 문자열은 'NONE'으로 치환되지 않음

✅ 올바른 예
```sql
SELECT PT_NAME,
       PT_NO,
       GEND_CD,
       AGE,
       COALESCE(NULLIF(TRIM(TLNO), ''), 'NONE') AS TLNO
FROM PATIENT
WHERE AGE <= 12 AND GEND_CD = 'W'
ORDER BY AGE DESC, PT_NAME ASC;
```

## 사용 개념 정리

- `TRIM(column)` : 문자열 앞뒤 공백 제거
- `NULLIF(expr1, expr2)` : `expr1 = expr2`이면 `NULL` 반환, 아니면 `expr1` 반환
- `COALESCE(expr1, expr2)` : `expr1`이 `NULL`이면 `expr2` 반환
- 결합하면 공백/빈 문자열/NULL 모두 `'NONE'`으로 처리 가능

````python
SELECT PT_NAME,
       PT_NO,
       GEND_CD,
       AGE,
       COALESCE(NULLIF(TRIM(TLNO), ''), 'NONE') AS TLNO
FROM PATIENT
WHERE AGE <= 12 AND GEND_CD = 'W'
ORDER BY AGE DESC, PT_NAME ASC;
````

## 한 줄 정리

> TLNO 컬럼을 처리할 때는 **NULL, 빈 문자열, 공백 문자열을 모두 고려해 `COALESCE(NULLIF(TRIM(...), ''), 'NONE')` 패턴을 사용**해야 한다.
