## 문제 요약

- 테이블: `BOOK`
- 조건:
  - `CATEGORY = '인문'`
  - 출판일이 **2021년도인 도서**만 조회
- 출력 조건:
  - 날짜를 `YYYY-MM-DD` (예: `2021-03-15`) 형태로 출력

`BOOK_ID`, `PUBLISHED_DATE`를 조회한다.

## 함정 / 실수 포인트

### ❗ LIKE로 날짜 비교하는 실수

- `PUBLISHED_DATE LIKE '2021%'`
- 날짜 컬럼에서는 문자열 비교가 되어 **권장되지 않음**

### ❗ 날짜 출력 형식 미변환

- 실제 저장 값이 `2021-03-15 00:00:00` 형태일 수 있음
- 문제에서 `YYYY-MM-DD`를 요구하면 **FORMAT 지정 필요**

## 사용 개념 정리

- `BETWEEN 날짜1 AND 날짜2`
  - 날짜 범위 필터링에 가장 안전한 방식
- `DATE_FORMAT(date, '%Y-%m-%d')`
  - 날짜 출력 형식 지정 (연-월-일)
- `WHERE` 절은 **집계 전 / 정렬 전**에 적용됨

```python
SELECT 
    BOOK_ID,
    DATE_FORMAT(PUBLISHED_DATE, '%Y-%m-%d') AS PUBLISHED_DATE
FROM BOOK
WHERE CATEGORY = '인문'
  AND PUBLISHED_DATE BETWEEN '2021-01-01' AND '2021-12-31'
ORDER BY PUBLISHED_DATE;
```

## 한 줄 정리

> 날짜 조건은 **BETWEEN**, 날짜 출력 형식은 **DATE_FORMAT('%Y-%m-%d')** 로 처리한다.
