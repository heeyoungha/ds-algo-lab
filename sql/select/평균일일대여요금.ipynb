## 문제 요약

- 테이블: `CAR_RENTAL_COMPANY_CAR`
- 조건: `CAR_TYPE = 'SUV'`
- 요구사항:
  - `DAILY_FEE`의 **평균**을 구한다
  - 결과를 **소수 첫째 자리에서 반올림**한다

최종 결과를 `AVERAGE_FEE` 컬럼으로 출력한다.

## 함정 / 실수 포인트

### ❗ AVG를 빼먹는 실수

- `ROUND(DAILY_FEE)` ❌
- 문제는 **평균값**을 요구함

### ❗ 반올림 기준 혼동

- `ROUND(x)` 또는 `ROUND(x, 0)` → 소수 첫째 자리에서 반올림
- 소수 둘째 자리 반올림이 아님에 주의

### ❗ 함수 적용 순서

- 반드시 `AVG → ROUND` 순서로 적용해야 함

## 사용 개념 정리

- `AVG(column)` : 평균을 구하는 집계 함수
- `ROUND(number)` : 소수 첫째 자리에서 반올림
- `WHERE` : 집계 전 조건 필터링
- 집계 함수는 **행 단위가 아닌 결과값에 적용**됨

## 정답 쿼리


```python
SELECT ROUND(AVG(DAILY_FEE)) AS AVERAGE_FEE
FROM CAR_RENTAL_COMPANY_CAR
WHERE CAR_TYPE = 'SUV';
```
